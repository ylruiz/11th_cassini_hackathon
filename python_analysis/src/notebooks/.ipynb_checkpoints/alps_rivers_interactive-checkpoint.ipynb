# Alpine Rivers Interactive Explorer

This notebook uses **ipywidgets controls** + **matplotlib** to choose AOI/time and refresh Sentinel-2 NDWI river maps.

1. Run all cells.
2. Adjust bbox/time/cloud/threshold.
3. Click **Download and Refresh**.

```python
from __future__ import annotations

from datetime import datetime
from pathlib import Path
from time import perf_counter
import hashlib
import json

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import openeo
import xarray as xr
import ipywidgets as widgets
from IPython.display import display, clear_output
from IPython import get_ipython

matplotlib.rcParams["text.usetex"] = False
matplotlib.rcParams["font.family"] = "sans-serif"
matplotlib.rcParams["font.sans-serif"] = ["DejaVu Sans", "Liberation Sans", "Arial"]
matplotlib.rcParams["mathtext.fontset"] = "dejavusans"

ip = get_ipython()
if ip is not None:
    try:
        ip.run_line_magic("matplotlib", "widget")
    except Exception:
        pass

def find_project_root(start: Path) -> Path:
    p = start.resolve()
    for candidate in [p] + list(p.parents):
        if (candidate / "data").exists() and (candidate / "src").exists():
            return candidate
    return p

ROOT = find_project_root(Path.cwd())
DATA_DIR = ROOT / "data"
DATA_DIR.mkdir(parents=True, exist_ok=True)

def validate_inputs(west: float, east: float, south: float, north: float, start: str, end: str) -> None:
    if not (west < east and south < north):
        raise ValueError("Invalid bbox: require west < east and south < north.")
    start_dt = datetime.strptime(start, "%Y-%m-%d")
    end_dt = datetime.strptime(end, "%Y-%m-%d")
    if start_dt > end_dt:
        raise ValueError("Invalid dates: start must be <= end.")

def cache_path_for_query(bbox: dict, start: str, end: str, cloud: float) -> Path:
    payload = {"bbox": bbox, "start": start, "end": end, "cloud": cloud}
    key = hashlib.md5(json.dumps(payload, sort_keys=True).encode("utf-8")).hexdigest()[:12]
    return DATA_DIR / f"alps_s2_{key}.nc"

def download_cube_if_needed(bbox: dict, start: str, end: str, cloud: float, force: bool, log) -> Path:
    nc_path = cache_path_for_query(bbox, start, end, cloud)
    if nc_path.exists() and not force:
        log(f"Using cache: {nc_path}")
        return nc_path

    log("Connecting to openEO backend...")
    conn = openeo.connect("openeo.dataspace.copernicus.eu").authenticate_oidc()

    log("Loading SENTINEL2_L2A (B03, B08)...")
    cube = conn.load_collection(
        "SENTINEL2_L2A",
        spatial_extent=bbox,
        temporal_extent=[start, end],
        bands=["B03", "B08"],
        max_cloud_cover=cloud,
    )

    log(f"Downloading to {nc_path} ...")
    cube.download(str(nc_path))
    log("Download done.")
    return nc_path

def compute_ndwi(nc_path: Path):
    ds = xr.open_dataset(nc_path)
    if "B03" not in ds.variables or "B08" not in ds.variables:
        raise RuntimeError("Missing B03/B08 variables in NetCDF.")

    green = ds["B03"].astype("float32")
    nir = ds["B08"].astype("float32")
    ndwi = (green - nir) / (green + nir)

    if "t" in ndwi.dims:
        ndwi2d = ndwi.median(dim="t", skipna=True)
    else:
        ndwi2d = ndwi

    arr = ndwi2d.values
    if not np.isfinite(arr).any():
        raise RuntimeError("No finite NDWI pixels. Try another AOI/date range.")

    lons = ndwi2d["x"].values.astype(float)
    lats = ndwi2d["y"].values.astype(float)
    return arr, lons, lats

def build_mask(arr: np.ndarray, threshold: float) -> np.ndarray:
    mask = np.zeros_like(arr, dtype=np.uint8)
    finite = np.isfinite(arr)
    mask[finite & (arr > threshold)] = 1
    return mask

print(f"Project root: {ROOT}")
print(f"Matplotlib backend: {matplotlib.get_backend()}")
```

```python
# Controls
west_w = widgets.FloatText(value=11.293602, description="west", layout=widgets.Layout(width="220px"))
east_w = widgets.FloatText(value=11.382866, description="east", layout=widgets.Layout(width="220px"))
south_w = widgets.FloatText(value=46.460163, description="south", layout=widgets.Layout(width="220px"))
north_w = widgets.FloatText(value=46.514768, description="north", layout=widgets.Layout(width="220px"))
start_w = widgets.Text(value="2023-07-01", description="start", layout=widgets.Layout(width="220px"))
end_w = widgets.Text(value="2023-07-20", description="end", layout=widgets.Layout(width="220px"))
cloud_w = widgets.FloatSlider(value=20.0, min=0, max=100, step=1, description="cloud %", readout_format=".0f", layout=widgets.Layout(width="360px"))
force_w = widgets.Checkbox(value=False, description="force download")
thr_w = widgets.FloatSlider(value=0.10, min=-0.2, max=0.6, step=0.01, description="NDWI thr", readout_format=".2f", layout=widgets.Layout(width="360px"))
refresh_btn = widgets.Button(description="Download and Refresh", button_style="primary")
status_w = widgets.HTML(value="<b>Status:</b> idle")

log_out = widgets.Output(layout=widgets.Layout(border="1px solid #bbb", height="180px", overflow="auto"))
plot_out = widgets.Output(layout=widgets.Layout(border="1px solid #bbb"))

state = {"arr": None, "lons": None, "lats": None}

def log(msg: str):
    with log_out:
        print(msg)

def render_plot():
    arr = state["arr"]
    lons = state["lons"]
    lats = state["lats"]
    if arr is None:
        return

    threshold = float(thr_w.value)
    mask = build_mask(arr, threshold)
    origin = "upper" if lats[0] > lats[-1] else "lower"
    extent = [float(lons.min()), float(lons.max()), float(lats.min()), float(lats.max())]

    with plot_out:
        clear_output(wait=True)
        fig, (ax_ndwi, ax_mask, ax_hist) = plt.subplots(1, 3, figsize=(14, 4.8))

        im = ax_ndwi.imshow(arr, cmap="RdBu", vmin=-1, vmax=1, origin=origin, extent=extent)
        ax_ndwi.set_title("NDWI")
        ax_ndwi.set_xlabel("Longitude")
        ax_ndwi.set_ylabel("Latitude")
        fig.colorbar(im, ax=ax_ndwi, shrink=0.85)

        ax_mask.imshow(mask, cmap="Blues", vmin=0, vmax=1, origin=origin, extent=extent)
        ax_mask.set_title(f"Water mask (NDWI > {threshold:.2f})")
        ax_mask.set_xlabel("Longitude")
        ax_mask.set_ylabel("Latitude")

        vals = arr[np.isfinite(arr)]
        ax_hist.hist(vals, bins=80, color="gray", alpha=0.85)
        ax_hist.axvline(threshold, color="dodgerblue", linestyle="--", linewidth=2)
        ax_hist.set_title("NDWI histogram")
        ax_hist.set_xlabel("NDWI")
        ax_hist.set_ylabel("Pixel count")

        plt.tight_layout()
        plt.show()

    total = int(np.isfinite(arr).sum())
    water = int(mask.sum())
    pct = 100.0 * water / total if total else 0.0
    status_w.value = f"<b>Status:</b> loaded | threshold={threshold:.2f} | water={water}/{total} ({pct:.2f}%)"

def on_threshold(change):
    if change["name"] == "value" and state["arr"] is not None:
        render_plot()

thr_w.observe(on_threshold, names="value")

def on_refresh(_):
    t0 = perf_counter()
    try:
        status_w.value = "<b>Status:</b> validating input..."
        west = float(west_w.value)
        east = float(east_w.value)
        south = float(south_w.value)
        north = float(north_w.value)
        start = start_w.value.strip()
        end = end_w.value.strip()
        cloud = float(cloud_w.value)
        force = bool(force_w.value)

        validate_inputs(west, east, south, north, start, end)
        bbox = {"west": west, "east": east, "south": south, "north": north, "crs": "EPSG:4326"}

        log(f"[{perf_counter()-t0:6.2f}s] Query bbox={bbox}, dates={start}..{end}, cloud<={cloud}%")
        status_w.value = "<b>Status:</b> downloading/opening data..."

        nc_path = download_cube_if_needed(bbox, start, end, cloud, force, log=log)
        arr, lons, lats = compute_ndwi(nc_path)

        state["arr"], state["lons"], state["lats"] = arr, lons, lats
        log(f"[{perf_counter()-t0:6.2f}s] NDWI stats min={np.nanmin(arr):.3f}, max={np.nanmax(arr):.3f}, mean={np.nanmean(arr):.3f}")

        render_plot()
        log(f"[{perf_counter()-t0:6.2f}s] Refresh done.")
    except Exception as exc:
        status_w.value = f"<b>Status:</b> error - {exc}"
        log(f"ERROR: {exc}")

refresh_btn.on_click(on_refresh)

left = widgets.VBox([west_w, east_w, south_w, north_w])
right = widgets.VBox([start_w, end_w, cloud_w, force_w, thr_w, refresh_btn, status_w])
controls = widgets.HBox([left, right])

display(controls)
display(plot_out)
display(widgets.HTML("<b>Logs</b>"))
display(log_out)

print("UI ready. Click 'Download and Refresh'.")
```
