# Oetztaler Glacier and Connected Rivers: Multi-Variable Trend Overview

This notebook targets the Oetztaler glacier area around `lat=46.83`, `lon=10.78` and builds a broad trend dashboard from openEO Data Space data.

Variables included:
- Sentinel-2 optical indices: `NDVI`, `NDWI`, `NDSI`
- Sentinel-1 radar signals: `VV(dB)`, `VH(dB)`, `VV-VH (dB)`
- Fraction proxies over AOI: snow/ice fraction (`NDSI > 0.4`) and water fraction (`NDWI > 0.1`)


```python
from __future__ import annotations

from pathlib import Path
from typing import Iterable

import matplotlib.pyplot as plt
import numpy as np
import openeo
import pandas as pd
import xarray as xr

%matplotlib widget

plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["axes.grid"] = True
```

```python
# Target AOI around Oetztaler gletscher
CENTER_LAT = 46.83
CENTER_LON = 10.78
HALF_SIZE_DEG = 0.08

START_DATE = "2018-01-01"
END_DATE = "2024-12-31"

MAX_CLOUD_COVER = 80
FORCE_DOWNLOAD = False

ROOT = Path("/home/vsilv/.nextcloud/src/cassini/11th_cassini_hackathon/python_analysis")
DATA_DIR = ROOT / "data" / "oetztal"
FIG_DIR = ROOT / "figures"
DATA_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)

S2_FILE = DATA_DIR / f"oetztal_s2_{START_DATE}_{END_DATE}.nc"
S1_FILE = DATA_DIR / f"oetztal_s1_{START_DATE}_{END_DATE}.nc"

BBOX = {
    "west": CENTER_LON - HALF_SIZE_DEG,
    "east": CENTER_LON + HALF_SIZE_DEG,
    "south": CENTER_LAT - HALF_SIZE_DEG,
    "north": CENTER_LAT + HALF_SIZE_DEG,
    "crs": "EPSG:4326",
}

print("AOI center:", CENTER_LAT, CENTER_LON)
print("AOI bbox:", BBOX)
print("Time range:", START_DATE, "to", END_DATE)
print("S2 file:", S2_FILE)
print("S1 file:", S1_FILE)
```

```python
def get_coord_name(ds: xr.Dataset, candidates: Iterable[str]) -> str:
    for c in candidates:
        if c in ds.coords or c in ds.dims:
            return c
    raise RuntimeError(f"None of {list(candidates)} found in coords/dims. Found coords={list(ds.coords)}, dims={list(ds.dims)}")


def get_time_name(ds: xr.Dataset) -> str:
    return get_coord_name(ds, ["t", "time", "date", "valid_time"])


def maybe_download() -> None:
    need_s2 = FORCE_DOWNLOAD or not S2_FILE.exists()
    need_s1 = FORCE_DOWNLOAD or not S1_FILE.exists()

    if not (need_s2 or need_s1):
        print("Using cached S1/S2 NetCDF files.")
        return

    print("Connecting to openEO backend...")
    conn = openeo.connect("openeo.dataspace.copernicus.eu").authenticate_oidc()

    if need_s2:
        print("Building Sentinel-2 L2A cube: B03,B04,B08,B11,SCL")
        s2_cube = conn.load_collection(
            "SENTINEL2_L2A",
            spatial_extent=BBOX,
            temporal_extent=[START_DATE, END_DATE],
            bands=["B03", "B04", "B08", "B11", "SCL"],
            max_cloud_cover=MAX_CLOUD_COVER,
        )
        print("Downloading S2 cube to", S2_FILE)
        s2_cube.download(str(S2_FILE))

    if need_s1:
        print("Building Sentinel-1 GRD cube: VV,VH")
        s1_cube = conn.load_collection(
            "SENTINEL1_GRD",
            spatial_extent=BBOX,
            temporal_extent=[START_DATE, END_DATE],
            bands=["VV", "VH"],
        )
        print("Downloading S1 cube to", S1_FILE)
        s1_cube.download(str(S1_FILE))


maybe_download()
```

```python
s2_ds = xr.open_dataset(S2_FILE, engine="netcdf4")
s1_ds = xr.open_dataset(S1_FILE, engine="netcdf4")

print("S2 dataset:")
print(s2_ds)
print("\nS1 dataset:")
print(s1_ds)
```

```python
# Normalize coord names
s2_t = get_time_name(s2_ds)
s2_y = get_coord_name(s2_ds, ["y", "lat", "latitude"])
s2_x = get_coord_name(s2_ds, ["x", "lon", "longitude"])

s1_t = get_time_name(s1_ds)
s1_y = get_coord_name(s1_ds, ["y", "lat", "latitude"])
s1_x = get_coord_name(s1_ds, ["x", "lon", "longitude"])

# Sentinel-2 cloud/snow/shadow mask from SCL
# Keep classes that are usually useful for land/water/ice monitoring:
# 4=veg, 5=bare soil, 6=water, 11=snow/ice
valid_scl = [4, 5, 6, 11]
scl = s2_ds["SCL"]
valid_mask = xr.zeros_like(scl, dtype=bool)
for v in valid_scl:
    valid_mask = valid_mask | (scl == v)

green = s2_ds["B03"].astype("float32")
red = s2_ds["B04"].astype("float32")
nir = s2_ds["B08"].astype("float32")
swir = s2_ds["B11"].astype("float32")

ndvi = ((nir - red) / (nir + red)).where(valid_mask)
ndwi = ((green - nir) / (green + nir)).where(valid_mask)
ndsi = ((green - swir) / (green + swir)).where(valid_mask)

# Monthly medians over AOI
def area_monthly_series(da: xr.DataArray, time_name: str) -> pd.Series:
    s = da.mean(dim=[d for d in da.dims if d != time_name], skipna=True).to_series()
    s.index = pd.to_datetime(s.index)
    return s.resample("MS").median()


ndvi_m = area_monthly_series(ndvi, s2_t)
ndwi_m = area_monthly_series(ndwi, s2_t)
ndsi_m = area_monthly_series(ndsi, s2_t)

# Fraction proxies (monthly): snow/ice and water share in AOI
snow_frac_m = area_monthly_series((ndsi > 0.4).astype(float), s2_t)
water_frac_m = area_monthly_series((ndwi > 0.1).astype(float), s2_t)

# Sentinel-1 monthly radar signals
vv = s1_ds["VV"].astype("float32")
vh = s1_ds["VH"].astype("float32")
eps = 1e-8
vv_db = 10.0 * np.log10(vv.clip(min=eps))
vh_db = 10.0 * np.log10(vh.clip(min=eps))
ratio_db = vv_db - vh_db

vv_m = area_monthly_series(vv_db, s1_t)
vh_m = area_monthly_series(vh_db, s1_t)
ratio_m = area_monthly_series(ratio_db, s1_t)

overview_df = pd.DataFrame({
    "NDVI": ndvi_m,
    "NDWI": ndwi_m,
    "NDSI": ndsi_m,
    "SnowFrac": snow_frac_m,
    "WaterFrac": water_frac_m,
    "VV_dB": vv_m,
    "VH_dB": vh_m,
    "VV_minus_VH_dB": ratio_m,
}).sort_index()

overview_df.tail(12)
```

```python
def fit_linear_trend(series: pd.Series) -> tuple[float, float, np.ndarray]:
    s = series.dropna()
    if len(s) < 3:
        return np.nan, np.nan, np.full(len(series), np.nan)
    x = s.index.year + (s.index.month - 0.5) / 12.0
    y = s.values.astype(float)
    m, b = np.polyfit(x, y, 1)

    full_x = series.index.year + (series.index.month - 0.5) / 12.0
    fit = m * full_x + b
    return float(m), float(b), fit


trend_info = {}
for col in overview_df.columns:
    m, b, fit = fit_linear_trend(overview_df[col])
    trend_info[col] = {
        "slope_per_year": m,
        "intercept": b,
        "fit": fit,
    }

trend_table = pd.DataFrame({k: {"slope_per_year": v["slope_per_year"]} for k, v in trend_info.items()}).T
trend_table
```

```python
fig, axes = plt.subplots(3, 1, figsize=(11, 12), sharex=True)

# Optical
for c in ["NDVI", "NDWI", "NDSI"]:
    axes[0].plot(overview_df.index, overview_df[c], marker="o", linewidth=1.2, label=c)
    axes[0].plot(overview_df.index, trend_info[c]["fit"], linestyle="--", linewidth=1.6, alpha=0.9)
axes[0].set_title("Optical index trends (monthly median over AOI)")
axes[0].set_ylabel("index value")
axes[0].legend(loc="best")

# Fractions
for c in ["SnowFrac", "WaterFrac"]:
    axes[1].plot(overview_df.index, overview_df[c], marker="o", linewidth=1.2, label=c)
    axes[1].plot(overview_df.index, trend_info[c]["fit"], linestyle="--", linewidth=1.6, alpha=0.9)
axes[1].set_title("AOI fraction proxies")
axes[1].set_ylabel("fraction of pixels")
axes[1].set_ylim(-0.05, 1.05)
axes[1].legend(loc="best")

# Radar
for c in ["VV_dB", "VH_dB", "VV_minus_VH_dB"]:
    axes[2].plot(overview_df.index, overview_df[c], marker="o", linewidth=1.2, label=c)
    axes[2].plot(overview_df.index, trend_info[c]["fit"], linestyle="--", linewidth=1.6, alpha=0.9)
axes[2].set_title("Radar signal trends")
axes[2].set_ylabel("dB")
axes[2].legend(loc="best")

plt.tight_layout()
plt.show()

fig.savefig(FIG_DIR / "oetztal_overview_timeseries_trends.png", dpi=160, bbox_inches="tight")
```

```python
# Monthly climatology (mean by month across all years)
clim = overview_df.copy()
clim["month"] = clim.index.month
clim_m = clim.groupby("month").mean(numeric_only=True)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for c in ["NDVI", "NDWI", "NDSI"]:
    axes[0].plot(clim_m.index, clim_m[c], marker="o", label=c)
axes[0].set_title("Optical monthly climatology")
axes[0].set_xlabel("month")
axes[0].set_ylabel("index value")
axes[0].legend()

for c in ["SnowFrac", "WaterFrac"]:
    axes[1].plot(clim_m.index, clim_m[c], marker="o", label=c)
axes[1].set_title("Fraction monthly climatology")
axes[1].set_xlabel("month")
axes[1].set_ylabel("fraction")
axes[1].set_ylim(-0.05, 1.05)
axes[1].legend()

for c in ["VV_dB", "VH_dB", "VV_minus_VH_dB"]:
    axes[2].plot(clim_m.index, clim_m[c], marker="o", label=c)
axes[2].set_title("Radar monthly climatology")
axes[2].set_xlabel("month")
axes[2].set_ylabel("dB")
axes[2].legend()

plt.tight_layout()
plt.show()

fig.savefig(FIG_DIR / "oetztal_overview_monthly_climatology.png", dpi=160, bbox_inches="tight")
```

```python
def pixel_slope_per_year(da: xr.DataArray, time_name: str) -> xr.DataArray:
    # Linear trend map: slope per year
    t = pd.to_datetime(da[time_name].values)
    x = xr.DataArray(t.year + (t.month - 0.5) / 12.0, dims=[time_name], coords={time_name: da[time_name]})

    x_mean = x.mean(time_name)
    y_mean = da.mean(time_name, skipna=True)
    cov = ((x - x_mean) * (da - y_mean)).mean(time_name, skipna=True)
    var = ((x - x_mean) ** 2).mean(time_name, skipna=True)
    return cov / var


ndsi_slope = pixel_slope_per_year(ndsi, s2_t)
ndwi_slope = pixel_slope_per_year(ndwi, s2_t)
ndvi_slope = pixel_slope_per_year(ndvi, s2_t)

vv_slope = pixel_slope_per_year(vv_db, s1_t)

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

im0 = axes[0, 0].imshow(ndsi_slope, cmap="RdBu_r")
axes[0, 0].set_title("NDSI slope per year")
plt.colorbar(im0, ax=axes[0, 0], shrink=0.8)

im1 = axes[0, 1].imshow(ndwi_slope, cmap="RdBu_r")
axes[0, 1].set_title("NDWI slope per year")
plt.colorbar(im1, ax=axes[0, 1], shrink=0.8)

im2 = axes[1, 0].imshow(ndvi_slope, cmap="RdBu_r")
axes[1, 0].set_title("NDVI slope per year")
plt.colorbar(im2, ax=axes[1, 0], shrink=0.8)

im3 = axes[1, 1].imshow(vv_slope, cmap="RdBu_r")
axes[1, 1].set_title("VV(dB) slope per year")
plt.colorbar(im3, ax=axes[1, 1], shrink=0.8)

for ax in axes.ravel():
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()

fig.savefig(FIG_DIR / "oetztal_overview_pixel_trend_maps.png", dpi=160, bbox_inches="tight")

print("Regional trend slopes per year:")
display(trend_table)
```
