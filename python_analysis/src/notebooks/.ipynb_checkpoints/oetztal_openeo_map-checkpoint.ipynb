# openEO map fetch (position + extent)

Interactive notebook to fetch and plot a map for any center position + extent.

Defaults are set for **Oetztal glacier** (`46.83, 10.78`).

```python
from __future__ import annotations

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import openeo
import xarray as xr

%matplotlib widget
```

```python
# -------- user parameters --------
BACKEND_URL = "openeo.dataspace.copernicus.eu"

CENTER_LAT = 46.83
CENTER_LON = 10.78
HALF_SIZE_DEG = 0.08  # extent control

START_DATE = "2024-06-01"
END_DATE = "2024-09-30"

COLLECTION_ID = "SENTINEL2_L2A"
BANDS = ["B02", "B03", "B04", "B08", "SCL"]
MAX_CLOUD_COVER = 70
FORCE_DOWNLOAD = False

ROOT = Path("/home/vsilv/.nextcloud/src/cassini/11th_cassini_hackathon/python_analysis")
DATA_DIR = ROOT / "data" / "oetztal"
FIG_DIR = ROOT / "figures"
DATA_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)

OUT_FILE = DATA_DIR / f"map_{COLLECTION_ID}_{START_DATE}_{END_DATE}_{CENTER_LAT:.3f}_{CENTER_LON:.3f}.nc"

BBOX = {
    "west": CENTER_LON - HALF_SIZE_DEG,
    "east": CENTER_LON + HALF_SIZE_DEG,
    "south": CENTER_LAT - HALF_SIZE_DEG,
    "north": CENTER_LAT + HALF_SIZE_DEG,
    "crs": "EPSG:4326",
}

print("Backend:", BACKEND_URL)
print("Collection:", COLLECTION_ID)
print("BBox:", BBOX)
print("Dates:", START_DATE, "to", END_DATE)
print("Output file:", OUT_FILE)
```

```python
def get_time_name(ds: xr.Dataset) -> str | None:
    for c in ["t", "time", "date", "valid_time"]:
        if c in ds.coords or c in ds.dims:
            return c
    return None


def normalize_rgb(arr: np.ndarray, q_low: float = 2.0, q_high: float = 98.0) -> np.ndarray:
    x = arr.astype(np.float32).copy()
    lo = np.nanpercentile(x, q_low)
    hi = np.nanpercentile(x, q_high)
    if not np.isfinite(lo) or not np.isfinite(hi) or hi <= lo:
        return np.zeros_like(x, dtype=np.float32)
    x = (x - lo) / (hi - lo)
    return np.clip(x, 0, 1)
```

```python
if OUT_FILE.exists() and not FORCE_DOWNLOAD:
    print("Using cached file:", OUT_FILE)
else:
    print("Connecting/authenticating...")
    conn = openeo.connect(BACKEND_URL).authenticate_oidc()

    print("Building cube...")
    cube = conn.load_collection(
        COLLECTION_ID,
        spatial_extent=BBOX,
        temporal_extent=[START_DATE, END_DATE],
        bands=BANDS,
        max_cloud_cover=MAX_CLOUD_COVER,
    )

    print("Downloading to", OUT_FILE)
    cube.download(str(OUT_FILE))
    print("Done.")
```

```python
ds = xr.open_dataset(OUT_FILE, engine="netcdf4")
print(ds)

tname = get_time_name(ds)
if tname:
    comp = ds.median(dim=tname, skipna=True)
else:
    comp = ds

# Optional SCL-based mask
if "SCL" in comp:
    valid = xr.zeros_like(comp["SCL"], dtype=bool)
    for v in [2, 4, 5, 6, 7, 11]:
        valid = valid | (comp["SCL"] == v)
else:
    valid = None

r = comp["B04"].astype("float32")
g = comp["B03"].astype("float32")
b = comp["B02"].astype("float32")

if valid is not None:
    r = r.where(valid)
    g = g.where(valid)
    b = b.where(valid)

rgb = np.dstack([
    normalize_rgb(r.values),
    normalize_rgb(g.values),
    normalize_rgb(b.values),
])

ndwi = ((comp["B03"].astype("float32") - comp["B08"].astype("float32")) /
        (comp["B03"].astype("float32") + comp["B08"].astype("float32")))
if valid is not None:
    ndwi = ndwi.where(valid)

water = (ndwi > 0.1).astype(np.uint8)

fig, ax = plt.subplots(1, 3, figsize=(15, 5), facecolor="white")

ax[0].imshow(rgb)
ax[0].set_title("RGB median composite")
ax[0].set_xticks([]); ax[0].set_yticks([])

cmap = plt.get_cmap("RdBu_r").copy()
cmap.set_bad("lightgray")
im1 = ax[1].imshow(ndwi, cmap=cmap)
ax[1].set_title("NDWI")
ax[1].set_xticks([]); ax[1].set_yticks([])
plt.colorbar(im1, ax=ax[1], shrink=0.8)

im2 = ax[2].imshow(water, cmap="Blues", vmin=0, vmax=1)
ax[2].set_title("Water mask (NDWI > 0.1)")
ax[2].set_xticks([]); ax[2].set_yticks([])
plt.colorbar(im2, ax=ax[2], shrink=0.8)

plt.tight_layout()
fig_path = FIG_DIR / "oetztal_openeo_map_rgb_ndwi_water.png"
fig.savefig(fig_path, dpi=160, bbox_inches="tight", facecolor="white")
plt.show()
print("Saved figure:", fig_path)
ds.close()
```
