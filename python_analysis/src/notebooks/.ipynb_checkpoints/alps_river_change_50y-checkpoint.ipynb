# Alpine River Change Analysis (Long-Term NDWI)

This notebook estimates river/water change over a long period using an NDWI proxy and a fixed AOI.

Notes:
- If true 50-year coverage is not available in the selected collection, the notebook skips missing years and continues.
- Output is static (%matplotlib inline), no widgets or HTML maps.

```python
from __future__ import annotations

from datetime import datetime
from pathlib import Path
from time import perf_counter
import hashlib
import json

import matplotlib
matplotlib.use("Agg", force=True)
import matplotlib.pyplot as plt
import numpy as np
import openeo
import xarray as xr

%matplotlib inline

matplotlib.rcParams["text.usetex"] = False
matplotlib.rcParams["font.family"] = "sans-serif"
matplotlib.rcParams["font.sans-serif"] = ["DejaVu Sans", "Liberation Sans", "Arial"]
matplotlib.rcParams["mathtext.fontset"] = "dejavusans"
```

```python
# Parameters (edit here)
AOI = {
    "west": 11.293602,
    "east": 11.382866,
    "south": 46.460163,
    "north": 46.514768,
    "crs": "EPSG:4326",
}

# Try long-term collection first. Adjust if needed.
COLLECTION_ID = "LANDSAT_BIMONTHLY_MOSAIC"
GREEN_BAND = "B03"
NIR_BAND = "B05"

START_YEAR = 1975
END_YEAR = datetime.now().year - 1

NDWI_THRESHOLD = 0.05
MAX_CLOUD_COVER = 40.0
FORCE_DOWNLOAD = False

def find_project_root(start: Path) -> Path:
    p = start.resolve()
    for candidate in [p] + list(p.parents):
        if (candidate / "data").exists() and (candidate / "src").exists():
            return candidate
    return p

ROOT = find_project_root(Path.cwd())
DATA_DIR = ROOT / "data" / "river_change_50y"
FIG_DIR = ROOT / "figures"
DATA_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)

print(f"ROOT: {ROOT}")
print(f"Collection: {COLLECTION_ID} ({GREEN_BAND}/{NIR_BAND})")
print(f"Years requested: {START_YEAR}..{END_YEAR}")
```

```python
def year_cache_path(collection_id: str, year: int, aoi: dict, cloud: float) -> Path:
    payload = {"collection": collection_id, "year": year, "aoi": aoi, "cloud": cloud}
    key = hashlib.md5(json.dumps(payload, sort_keys=True).encode("utf-8")).hexdigest()[:12]
    return DATA_DIR / f"{collection_id}_{year}_{key}.nc"

def load_or_download_year(
    conn,
    collection_id: str,
    year: int,
    aoi: dict,
    green_band: str,
    nir_band: str,
    max_cloud_cover: float,
    force_download: bool,
) -> Path:
    nc_path = year_cache_path(collection_id, year, aoi, max_cloud_cover)
    if nc_path.exists() and not force_download:
        return nc_path

    cube = conn.load_collection(
        collection_id,
        spatial_extent=aoi,
        temporal_extent=[f"{year}-01-01", f"{year}-12-31"],
        bands=[green_band, nir_band],
        max_cloud_cover=max_cloud_cover,
    )
    cube.download(str(nc_path))
    return nc_path

def compute_ndwi_from_file(nc_path: Path, green_band: str, nir_band: str):
    ds = xr.open_dataset(nc_path)
    if green_band not in ds.variables or nir_band not in ds.variables:
        raise RuntimeError(f"Missing expected bands {green_band}/{nir_band} in {nc_path.name}")

    g = ds[green_band].astype("float32")
    n = ds[nir_band].astype("float32")
    ndwi = (g - n) / (g + n)
    ndwi2d = ndwi.median(dim="t", skipna=True) if "t" in ndwi.dims else ndwi

    arr = ndwi2d.values
    if not np.isfinite(arr).any():
        raise RuntimeError("NDWI result has no finite pixels")

    return arr, ndwi2d["x"].values.astype(float), ndwi2d["y"].values.astype(float)

def build_mask(ndwi_arr: np.ndarray, threshold: float) -> np.ndarray:
    mask = np.zeros_like(ndwi_arr, dtype=np.uint8)
    finite = np.isfinite(ndwi_arr)
    mask[finite & (ndwi_arr > threshold)] = 1
    return mask
```

```python
# Run yearly analysis
t0 = perf_counter()

conn = openeo.connect("openeo.dataspace.copernicus.eu").authenticate_oidc()
available_ids = set(conn.list_collection_ids())
if COLLECTION_ID not in available_ids:
    raise ValueError(f"Collection {COLLECTION_ID} not available. Available count: {len(available_ids)}")

years = list(range(START_YEAR, END_YEAR + 1))
results = []
maps = {}
lons = None
lats = None

for year in years:
    try:
        nc_path = load_or_download_year(
            conn=conn,
            collection_id=COLLECTION_ID,
            year=year,
            aoi=AOI,
            green_band=GREEN_BAND,
            nir_band=NIR_BAND,
            max_cloud_cover=MAX_CLOUD_COVER,
            force_download=FORCE_DOWNLOAD,
        )
        ndwi_arr, lons_y, lats_y = compute_ndwi_from_file(nc_path, GREEN_BAND, NIR_BAND)
        mask = build_mask(ndwi_arr, NDWI_THRESHOLD)

        finite = int(np.isfinite(ndwi_arr).sum())
        water = int(mask.sum())
        frac = water / finite if finite else np.nan

        results.append((year, frac, finite, water))
        maps[year] = mask
        lons = lons_y
        lats = lats_y
        print(f"{year}: water_fraction={frac:.4f} ({water}/{finite})")
    except Exception as exc:
        print(f"{year}: skipped ({exc})")

if not results:
    raise RuntimeError("No valid years processed. Try another collection/bands/AOI/date range.")

years_ok = np.array([r[0] for r in results], dtype=int)
water_frac = np.array([r[1] for r in results], dtype=float)

print(f"\nProcessed years: {len(results)} / {len(years)}")
print(f"Elapsed: {perf_counter() - t0:.2f}s")
```

```python
# Time-series plot of annual water fraction
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(years_ok, water_frac, marker="o", linewidth=1.8, color="tab:blue")

if len(years_ok) >= 3:
    coeff = np.polyfit(years_ok, water_frac, 1)
    trend = np.polyval(coeff, years_ok)
    ax.plot(years_ok, trend, linestyle="--", color="tab:red", label=f"trend: {coeff[0]:+.3e} / year")
    ax.legend(loc="best")

ax.set_title("Annual water fraction (NDWI proxy)")
ax.set_xlabel("Year")
ax.set_ylabel("Water pixels / valid pixels")
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Spatial comparison: earliest vs latest valid year
y0 = int(years_ok.min())
y1 = int(years_ok.max())
m0 = maps[y0]
m1 = maps[y1]
diff = m1.astype(int) - m0.astype(int)  # -1 lost water, +1 gained water

origin = "upper" if lats[0] > lats[-1] else "lower"
extent = [float(lons.min()), float(lons.max()), float(lats.min()), float(lats.max())]

fig, axes = plt.subplots(1, 3, figsize=(14, 5), sharex=True, sharey=True)

axes[0].imshow(m0, cmap="Blues", vmin=0, vmax=1, origin=origin, extent=extent)
axes[0].set_title(f"Water mask {y0}")

axes[1].imshow(m1, cmap="Blues", vmin=0, vmax=1, origin=origin, extent=extent)
axes[1].set_title(f"Water mask {y1}")

im = axes[2].imshow(diff, cmap="bwr", vmin=-1, vmax=1, origin=origin, extent=extent)
axes[2].set_title(f"Change {y0} -> {y1}")

for ax in axes:
    ax.set_xlabel("Longitude")
axes[0].set_ylabel("Latitude")

cbar = fig.colorbar(im, ax=axes.ravel().tolist(), shrink=0.85)
cbar.set_label("-1 lost water | +1 gained water")

plt.tight_layout()
plt.show()
```
